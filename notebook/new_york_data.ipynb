```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

print("="* 80)
print("NYC AIRBNB DATA EXPLORATION")
print("="* 80 )

#Load Data
print("\n Loading Data...")
listings = pd.read_csv("/Users/subhashmothukurigmail.com/Projects/ML_stuff/data/airbnb/listings.csv")
reviews = pd.read_csv("/Users/subhashmothukurigmail.com/Projects/ML_stuff/data/airbnb/reviews.csv")


print(f"Listing: {listings.shape[0]:,}rows, {listings.shape[1]} columns")
print(f"Reviews: {reviews.shape[0]:,} rows, {reviews.shape[1]} columns")

#Listing Analysis
print("\n" + "="*80)
print("LISTINGS.CSV ANALYSIS")
print("="*80)

print(f"\n📋 ALL COLUMNS ({listings.shape[1]} total):")
for i, col in enumerate(listings.columns, 1):
    dtype = listings[col].dtype
    nulls = listings[col].isnull().sum()
    null_pct = (nulls / len(listings)) * 100
    
    if nulls == len(listings):
        status = "❌ DEAD (100% null)"
    elif nulls > len(listings) * 0.5:
        status = "⚠️  MOSTLY NULL (>50%)"
    elif nulls == 0:
        status = "✅ COMPLETE"
    else:
        status = f"⚠️  {null_pct:.1f}% null"
    
    print(f"   {i:2d}. {col:40s} | {str(dtype):15s} | {status}")
```
